# Task 1: Data Cleaning & Preprocessing


```python
import pandas as pd
import numpy as np

pd.set_option('display.max_columns', None)

df = pd.read_csv('titanic_raw.csv')
print("Shape:", df.shape)
df.head()
```

## 1. Initial Inspection

```python
df.info()
```

```python
df.describe(include='all')
```

```python
print("Missing values per column:")
print(df.isnull().sum())
print("\nDuplicate rows:", df.duplicated().sum())
```

## 2. Handle Missing Values


```python
df['Age'] = df['Age'].fillna(df['Age'].median())
df['Embarked'] = df['Embarked'].fillna(df['Embarked'].mode()[0])
df['Fare'] = df['Fare'].fillna(df['Fare'].median())

df['Has_Cabin'] = df['Cabin'].notnull().astype(int)
df = df.drop(columns=['Cabin'])

print("Remaining missing values:\n", df.isnull().sum())
```

## 3. Remove Duplicates

```python
before = len(df)
df = df.drop_duplicates()
after = len(df)
print(f"Removed {before - after} duplicate rows. New shape: {df.shape}")
```

## 4. Fix Inconsistent Values & Convert Data Types


```python
df['Sex'] = df['Sex'].str.strip().str.lower()
print(df['Sex'].unique())

df['Pclass'] = df['Pclass'].astype('category')
df['Sex'] = df['Sex'].astype('category')
df['Embarked'] = df['Embarked'].astype('category')
df['Survived'] = df['Survived'].astype('int')

df.dtypes
```

## 5. Rename Columns (clean, consistent snake_case)

```python
df = df.rename(columns={
    'PassengerId': 'passenger_id',
    'Survived': 'survived',
    'Pclass': 'pclass',
    'Name': 'name',
    'Sex': 'sex',
    'Age': 'age',
    'SibSp': 'siblings_spouses',
    'Parch': 'parents_children',
    'Fare': 'fare',
    'Embarked': 'embarked',
    'Has_Cabin': 'has_cabin'
})
df.columns.tolist()
```

## 6. Final Check & Save Clean Dataset

```python
print("Final shape:", df.shape)
print("Missing values:\n", df.isnull().sum().sum(), "total")
df.head()
```

```python
df.to_csv('titanic_cleaned.csv', index=False)
print("Saved cleaned dataset -> titanic_cleaned.csv")
```
